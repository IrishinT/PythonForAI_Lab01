Напомним, что в соответствии с григорианским календарем, год является високосным, если его номер кратен 4, но не кратен 100, а также если он кратен 400.

````python
# ============================================================
# Задание 1: Високосный год (0.5 балла)
# ============================================================
# ОПИСАНИЕ:
# Определить, является ли год високосным по григорианскому календарю.
#
# УСЛОВИЯ ВИСОКОСНОСТИ:
# - Год кратен 4, но не кратен 100
# - ИЛИ год кратен 400
#
# ПРИМЕРЫ:
# 2020 -> YES (кратен 4, не кратен 100)
# 1900 -> NO (кратен 100, но не кратен 400)
# 2000 -> YES (кратен 400)
# ============================================================
````

````python
"""
Определяет високосный год по его номеру.

Аргументы:
    year (int) = номер года

Возвращает:
    str = Является ли год високосным "YES"/"NO"

"""

def isYearLeap(year):
    if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
        return "YES"
    return "NO"


# Тесты
def test_isYearLeap():
    assert isYearLeap(2018) == "NO"
    assert isYearLeap(2004) == "YES"
    assert isYearLeap(2020) == "YES"
    assert isYearLeap(2000) == "YES"
    assert isYearLeap(1984) == "YES"
    assert isYearLeap(1) == "NO"
    assert isYearLeap(4) == "YES"
    assert isYearLeap(100) == "NO"
    assert isYearLeap(400) == "YES"
    assert isYearLeap(1600) == "YES"
    assert isYearLeap(1700) == "NO"
    assert isYearLeap(1800) == "NO"
    assert isYearLeap(1900) == "NO"
    assert isYearLeap(2016) == "YES"
    assert isYearLeap(2024) == "YES"
    assert isYearLeap(2019) == "NO"
    assert isYearLeap(2021) == "NO"
    assert isYearLeap(2022) == "NO"
    assert isYearLeap(2023) == "NO"
    print("Все тесты пройдены!")

test_isYearLeap()
````

__Задание 2 (0.5 балла):__ Дано натуральное число. Найдите число знаков в его десятичной записи. Предложите как минимум два различных решения.

````python
# ============================================================
# Задание 2: Число знаков в десятичной записи (0.5 балла)
# ============================================================
# ОПИСАНИЕ:
# Найти количество цифр в натуральном числе.
#
# ПРИМЕРЫ:
# 1 -> 1
# 123 -> 3
# 1000 -> 4
# 987654321 -> 9
# ============================================================
````

````python
def countDigitsByStr(num):
    """
    Подсчитывает количество цифр благодаря преобразованию числа к строке и подсчёту количеству символов в ней через встроенную функцию len(str)

    Аргументы:
        num (int) - Число для подсчёта

    Возвращает:
        int - Количество цифр в десятичной записи

    """
    return len(str(num))

def countDigitsByDivide(num):
    """
    Подсчитывает количество цифр методом последовательного деления на основание (10)

    Аргументы:
        num (int) - Число для подсчёта

    Возвращает:
        int - Количество цифр в десятичной записи

    """
    if num == 0:
        return 1

    count = 0
    while num > 0:
        num //= 10
        count += 1

    return count


# Тесты

def test_countDigits(func):
    """
    Универсальная функция тестирования для подсчета цифр

    Аргументы:
        func (callable): тестируемая функция подсчёта цифр
    """
    print(f"Тестирование {func.__name__}:")

    assert func(0) == 1
    assert func(1) == 1
    assert func(9) == 1
    assert func(10) == 2
    assert func(99) == 2
    assert func(100) == 3
    assert func(999) == 3
    assert func(1000) == 4
    assert func(1234567890) == 10
    assert func(10**6) == 7
    assert func(10**9) == 10

    print("  Все тесты пройдены!")

test_countDigits(countDigitsByStr)
test_countDigits(countDigitsByDivide)
````

__Задание 3 (1 балл):__ По данному натуральном n вычислите сумму 1!+2!+3!+...+n!. В решении этой задачи с помощью циклов можно использовать только один цикл. Предложите как минимум два различных решения.

````python
# ============================================================
# Задание 3: Сумма факториалов (1 балл)
# ============================================================
# ОПИСАНИЕ:
# Вычислить сумму 1! + 2! + 3! + ... + n!
#
# ПРИМЕРЫ:
# n=1 -> 1
# n=3 -> 1!+2!+3! = 1+2+6 = 9
# n=5 -> 1+2+6+24+120 = 153
# ============================================================
````

````python
def task_03_by_recursion(n):
    """
    Вычисляет сумму факториала числа n с помощью рекурсивного метода

    Аргументы:
        num (int) - Число для которого нужно определить сумму факториалов 1! + 2! + 3! + ... + n!

    Возвращает:
        int - Итоговая сумма

    """

    if n <= 0:
        return 0

    def factor(k):
        if k <= 1:
            return 1

        return k * factor(k - 1)

    if n == 1:
        return 1

    return factor(n) + task_03_by_recursion(n - 1)

def task_03_by_cycle(n):
    """
    Вычисляет сумму факториала числа n с помощью простого цикла

    Аргументы:
        num (int) - Число для которого нужно определить сумму факториалов 1! + 2! + 3! + ... + n!

    Возвращает:
        int - Итоговая сумма

    """
    if n <= 0:
        return 0

    count = 0
    factor = 1

    for i in range(1, n+1):
        factor *= i
        count += factor

    return count

def test_factorial(func):
    """
    Универсальная функция тестирования для суммы факториалов

    Аргументы:
        func (callable): тестируемая функция
    """
    print(f"Тестирование {func.__name__}:")

    assert func(0) == 0
    assert func(1) == 1
    assert func(2) == 3
    assert func(3) == 9
    assert func(4) == 33
    assert func(5) == 153
    assert func(6) == 873
    assert func(-1) == 0
    assert func(-10) == 0

    print("  Все тесты пройдены!")

test_factorial(task_03_by_recursion)
test_factorial(task_03_by_cycle)
````

````python
def task_04_reverse(s):
    """
    Вычисляет является ли строка палидромом благодаря приведению строк к нижнему регистру и переворачиваю строки с помощью среза с шагом -1.

    Аргументы:
        s (str) - строка для проверки

    Возвращает:
        bool - является ли строка палидромом

    """
    return s.lower()[::-1] == s.lower()

def task_04_test():
    assert task_04_reverse("шалаш")
    assert task_04_reverse("казак")
    assert task_04_reverse("Казак")
    assert task_04_reverse("ШалаШ")
    assert task_04_reverse("дед")
    assert task_04_reverse("заказ")
    assert task_04_reverse("потоп")
    assert task_04_reverse("анна")
    assert not task_04_reverse("привет")
    assert not task_04_reverse("АБВГД")
    assert not task_04_reverse("Привет")
    assert not task_04_reverse("Hello")
    assert not task_04_reverse("World")
    assert task_04_reverse("12345654321")
    assert task_04_reverse("123321")
    assert task_04_reverse("abc123321cba")
    assert task_04_reverse("")
    assert task_04_reverse("a")

    print('Все тесты пройдены')

task_04_test()
````

Считать, что слова разделяются пробелами. Предложите как минимум два различных решения.

````python
# ============================================================
# Задание 5: Частота слов (1 балл)
# ============================================================
# ОПИСАНИЕ:
# Подсчитать, сколько раз каждое слово встречается в тексте.
# Слова разделяются пробелами.
#
#
# ПРИМЕРЫ:
# "hello world hello" -> {"hello": 2, "world": 1}
# "a a a" -> {"a": 3}
# "" -> {}
# ============================================================
````

````python
from collections import Counter

def task_05_countWordsInTextWithMapAndCycle(text):
    """
    Вычисляет количество слов (групп символов разделённых пробелом) в тексте, создавая словарь и обходя циклом весь текст подсчитывая слова вручную.

    Аргументы:
        text (str) - текст для подсчёта слов

    Возвращает:
        map - словарь с слово:количество - результат работы функции

    """
    words = text.split()

    wordCountMap = {}

    for word in words:
        wordCountMap[word.lower()] = wordCountMap.get(word.lower(), 0) + 1

    return wordCountMap

def task_05_countWordsInTextWithCounter(text):
    """
    Вычисляет количество слов (групп символов разделённых пробелом) в тексте с использованием библиотечного класса Counter

    Аргументы:
        text (str) - текст для подсчёта слов

    Возвращает:
        map - словарь с слово:количество - результат работы функции

    """
    words = list()

    for word in text.split():
        words.append(word.lower())

    return dict(Counter(words))

def task_05_test(func):
    """
    Универсальная функция тестирования для подсчёта слов

    Аргументы:
        func (callable): тестируемая функция
    """
    print(f"Тестирование {func.__name__}:")

    assert func("hello world hello") == {"hello": 2, "world": 1}
    assert func("a a a") == {"a": 3}
    assert func("") == {}
    assert func("кот котенок кот") == {'кот': 2, 'котенок': 1}
    assert func("кот кот кот") == {'кот': 3}
    assert func("код кот кот") == {'кот': 2, 'код': 1}
    assert func("Код кот кот") == {'кот': 2, 'код': 1}
    assert func("Клод код клод") == {'клод': 2, 'код': 1}
    assert func("Привет мир") == {'привет': 1, 'мир': 1}
    assert func("hello") == {"hello": 1}
    assert func("world") == {"world": 1}
    assert func("a b c") == {"a": 1, "b": 1, "c": 1}
    assert func("cat dog bird") == {"cat": 1, "dog": 1, "bird": 1}
    assert func("один два три") == {"один": 1, "два": 1, "три": 1}
    assert func("123 456 123") == {"123": 2, "456": 1}
    assert func("1 2 3 2 1") == {"1": 2, "2": 2, "3": 1}

    print("Все тесты пройдены!")

task_05_test(task_05_countWordsInTextWithMapAndCycle)
task_05_test(task_05_countWordsInTextWithCounter)
````

- если ни разу - кортеж (None, None).

Запрещается делать более одного прохода по каждому элементу строки.

````python
# ============================================================
# Задание 6: Поиск символа в строке (1 балл)
# ============================================================
# ОПИСАНИЕ:
# Найти первое и последнее вхождение символа в строке.
#
# ВОЗВРАЩАЕМЫЕ ЗНАЧЕНИЯ:
# - 0 вхождений: (None, None)
# - 1 вхождение: (индекс, None)
# - 2+ вхождений: (индекс_первого, индекс_последнего)
#
# ПРИМЕРЫ:
# "hello", "l" -> (2, 3)
# "hello", "h" -> (0, None)
# "hello", "x" -> (None, None)
# ============================================================
````

````python
def task_06_getCharPositions(input_str, input_char):
    """
    Ищет первое и последнее вхождение символа в строке.

    Аргументы:
        input_str (str) - текст в котором производится поиск
        input_str (char) - символ который ищем в тексте

    Возвращает:
        cottage = (первая позиция, последняя позиция) - позиции считаются с нуля.

    """


    startP = None
    endP = None

    for i in range(0, len(input_str)):
        char = input_str[i]

        if char == input_char:

            if startP is None:
                startP = i
            else:
                endP = i


    return (startP, endP)

def task_06_test():
    assert task_06_getCharPositions("hello", "l") == (2, 3)
    assert task_06_getCharPositions("hello", "h") == (0, None)
    assert task_06_getCharPositions("hello", "x") == (None, None)
    assert task_06_getCharPositions("world", "w") == (0, None)
    assert task_06_getCharPositions("world", "d") == (4, None)
    assert task_06_getCharPositions("world", "r") == (2, None)
    assert task_06_getCharPositions("aaa", "a") == (0, 2)
    assert task_06_getCharPositions("ababa", "a") == (0, 4)
    assert task_06_getCharPositions("ababa", "b") == (1, 3)
    assert task_06_getCharPositions("aaaaa", "a") == (0, 4)
    assert task_06_getCharPositions("12345", "1") == (0, None)
    assert task_06_getCharPositions("12345", "3") == (2, None)
    assert task_06_getCharPositions("12345", "5") == (4, None)
    assert task_06_getCharPositions("12121", "1") == (0, 4)
    assert task_06_getCharPositions("12121", "2") == (1, 3)
    assert task_06_getCharPositions("привет", "п") == (0, None)
    assert task_06_getCharPositions("привет", "т") == (5, None)
    assert task_06_getCharPositions("привет", "и") == (2, None)


    print('Все тесты пройдены')

task_06_test()

````
