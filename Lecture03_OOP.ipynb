В процедурном - выносим отдельные блоки программы в функции (функции не ограничиваются как в функци-м)
ООП - методология прогр основанная на представлении программы в виде совокупности объектов, каждый из которых является экземпляром определённого рода класса, а класы образуют иерархию наследования

Классы в пайтоне представляют из себя шаблоны или определения для создания объектов определённого типа

Магические методы (служебные)
__class__ - получить название класса
__str__ - аналог toString()

```python
class Person:

    @property
    def name(self):
        return self._name

    @property
    def height(self):
        return self._height

    @name.setter
    def name(self, name):
        self._name = name

    @height.deleter
    def height(self):
        del self.height

    def countIMT(self, weight):
        return round(weight / (self._height/100)**2, 2)

    @property
    def minimum_steps(self):
        return self._minimum_steps

    @minimum_steps.setter
    def minimum_steps(self, minimum_steps):
        self.__minimumsteps = minimum_steps

    @classmethod
    def default_minimum_steps(cls, minimum_steps):
        print(cls)
        type(cls)
        minimum_steps = cls(minimum_steps, "__minimum_steps", None)

    @staticmethod
    def count_steps(today_steps, minimum_steps=10000):
        if today_steps < minimum_steps:
            return "Физическая нагрузка на сегодня недостаточна"

        return "Отличная работа"

    def __init__(self, name, age, height):
        self._name = name
        self._age = age
        self._height = height
        pass

    def __str__(self):
        return f"Объект {self.__class__} с полями: имя {self._name}, возраст {self._age}, рост {self._height}"


vasya = Person("Вася", 43, 200)
print(vasya)
print(vasya.countIMT(weight=100))
print(Person.count_steps(100, 200))
print(vasya.count_steps(100, 200))
```
